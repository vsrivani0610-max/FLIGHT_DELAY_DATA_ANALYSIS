### Perform Mathematical Operations on Flight Data

```python
numerical_cols = [
    'DepTime', 'ArrTime', 'CRSArrTime', 'FlightNum', 'ActualElapsedTime',
    'CRSElapsedTime', 'AirTime', 'ArrDelay', 'DepDelay', 'Distance',
    'TaxiIn', 'TaxiOut', 'Cancelled', 'Diverted', 'CarrierDelay',
    'WeatherDelay', 'NASDelay', 'SecurityDelay', 'LateAircraftDelay'
]

# Create a NumPy array with only numerical data
flight_data_numeric = df[numerical_cols].to_numpy()

print(f"Shape of the numerical NumPy array: {flight_data_numeric.shape}")
```

```python
# Get the index of 'ArrDelay' and 'DepDelay' columns in the numerical_cols list
arr_delay_idx = numerical_cols.index('ArrDelay')
dep_delay_idx = numerical_cols.index('DepDelay')

# Extract 'ArrDelay' and 'DepDelay' columns from the numerical array
arr_delay_data = flight_data_numeric[:, arr_delay_idx]
dep_delay_data = flight_data_numeric[:, dep_delay_idx]

# Convert to float to ensure correct numerical operations, if not already.
# In this dataset, they are already int64, but this is a good practice.
arr_delay_data = arr_delay_data.astype(float)
dep_delay_data = dep_delay_data.astype(float)

print("\n--- Arrival Delay (ArrDelay) Statistics ---")
print(f"Mean Arrival Delay: {np.mean(arr_delay_data):.2f}")
print(f"Sum of Arrival Delays: {np.sum(arr_delay_data):.2f}")
print(f"Minimum Arrival Delay: {np.min(arr_delay_data):.2f}")
print(f"Maximum Arrival Delay: {np.max(arr_delay_data):.2f}")
print(f"Standard Deviation of Arrival Delays: {np.std(arr_delay_data):.2f}")

print("\n--- Departure Delay (DepDelay) Statistics ---")
print(f"Mean Departure Delay: {np.mean(dep_delay_data):.2f}")
print(f"Sum of Departure Delays: {np.sum(dep_delay_data):.2f}")
print(f"Minimum Departure Delay: {np.min(dep_delay_data):.2f}")
print(f"Maximum Departure Delay: {np.max(dep_delay_data):.2f}")
print(f"Standard Deviation of Departure Delays: {np.std(dep_delay_data):.2f}")
```

### Perform Indexing and Slicing on Flight Data

```python
# Accessing a specific element (e.g., 'ArrDelay' of the first flight)
# Remember 'ArrDelay' is at index 7 in `numerical_cols`
first_flight_arr_delay = flight_data_numeric[0, numerical_cols.index('ArrDelay')]
print(f"Arrival Delay of the first flight: {first_flight_arr_delay}")

# Accessing a specific row (e.g., data for the 5th flight)
fifth_flight_data = flight_data_numeric[4, :]
print(f"\nData for the 5th flight (all numerical columns):\n{fifth_flight_data}")

# Accessing a specific column (e.g., all 'Distance' values)
distance_idx = numerical_cols.index('Distance')
all_distances = flight_data_numeric[:, distance_idx]
print(f"\nFirst 5 'Distance' values: {all_distances[:5]}")
```

```python
# Slicing rows (e.g., data for the first 10 flights)
first_ten_flights = flight_data_numeric[0:10, :]
print(f"\nShape of data for the first 10 flights: {first_ten_flights.shape}")

# Slicing columns (e.g., 'DepTime' to 'FlightNum' for all flights)
dep_time_idx = numerical_cols.index('DepTime')
flight_num_idx = numerical_cols.index('FlightNum')

dep_time_to_flight_num = flight_data_numeric[:, dep_time_idx : flight_num_idx + 1]
print(f"\nShape of 'DepTime' to 'FlightNum' columns for all flights: {dep_time_to_flight_num.shape}")
print(f"First 3 rows of 'DepTime' to 'FlightNum':\n{dep_time_to_flight_num[:3]}")

# Slicing both rows and columns (e.g., 'ArrDelay' and 'DepDelay' for the first 5 flights)
subset_delays = flight_data_numeric[0:5, [numerical_cols.index('ArrDelay'), numerical_cols.index('DepDelay')]]
print(f"\n'ArrDelay' and 'DepDelay' for the first 5 flights:\n{subset_delays}")
```

### Boolean Indexing

Boolean indexing allows you to select elements from an array based on a condition. For example, we can find all flights where the arrival delay was greater than 100 minutes.

```python
# Find flights where 'ArrDelay' is greater than 100 minutes
arr_delay_idx = numerical_cols.index('ArrDelay')

# Create a boolean mask
long_arr_delay_mask = flight_data_numeric[:, arr_delay_idx] > 100

# Apply the mask to get the rows that satisfy the condition
long_delay_flights = flight_data_numeric[long_arr_delay_mask]

print(f"\nNumber of flights with arrival delay > 100 minutes: {len(long_delay_flights)}")
print(f"First 5 flights with arrival delay > 100 minutes (all numerical columns):\n{long_delay_flights[:5]}")

# You can also select specific columns for these filtered rows
long_delay_arr_dep = long_delay_flights[:, [numerical_cols.index('ArrDelay'), numerical_cols.index('DepDelay')]]
print(f"\nFirst 5 'ArrDelay' and 'DepDelay' values for flights with long arrival delays:\n{long_delay_arr_dep[:5]}")
```

### Histograms of Numerical Columns

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

### Histograms for Selected Numerical Columns

```python
selected_columns_to_plot = ['ArrDelay', 'DepDelay', 'Distance']

plt.figure(figsize=(15, 5))

for i, col_name in enumerate(selected_columns_to_plot):
    # Get the index of the column in the numerical_cols list
    col_index = numerical_cols.index(col_name)

    plt.subplot(1, len(selected_columns_to_plot), i + 1)
    sns.histplot(flight_data_numeric[:, col_index], kde=True, bins=30)
    plt.title(f'Distribution of {col_name}')
    plt.xlabel(col_name)
    plt.ylabel('Frequency')

plt.tight_layout()
plt.show()
```

### Histograms (Count Plots) of Categorical Columns

```python
categorical_cols_to_plot = ['DayOfWeek', 'UniqueCarrier', 'Airline', 'Origin', 'Dest', 'CancellationCode']

n_cols = 3 # Number of columns for the subplot grid
n_rows = (len(categorical_cols_to_plot) + n_cols - 1) // n_cols # Calculate required rows

plt.figure(figsize=(n_cols * 6, n_rows * 5))
plt.suptitle('Distribution of Categorical Flight Data Columns', y=1.02, fontsize=16)

for i, col_name in enumerate(categorical_cols_to_plot):
    plt.subplot(n_rows, n_cols, i + 1)
    sns.countplot(data=df, x=col_name, palette='viridis', order=df[col_name].value_counts().index)
    plt.title(f'Distribution of {col_name}')
    plt.xlabel(col_name)
    plt.ylabel('Count')
    plt.xticks(rotation=45, ha='right') # Rotate labels for better readability

plt.tight_layout(rect=[0, 0.03, 1, 0.98]) # Adjust layout to prevent titles overlapping
plt.show()
```

### Bivariate Analysis: Exploring Relationships Between Variables

#### 1. Numerical vs. Numerical: Scatter Plots

Let's visualize the relationship between Departure Delay and Arrival Delay, and also between Distance and Air Time.

```python
plt.figure(figsize=(12, 5))

# Scatter Plot: Departure Delay vs. Arrival Delay
plt.subplot(1, 2, 1)
sns.scatterplot(data=df, x='DepDelay', y='ArrDelay', alpha=0.5, hue='DayOfWeek', palette='viridis', size='Distance', sizes=(20, 200))
plt.title('Departure Delay vs. Arrival Delay')
plt.xlabel('Departure Delay (minutes)')
plt.ylabel('Arrival Delay (minutes)')
plt.grid(True, linestyle='--', alpha=0.6)

# Scatter Plot: Distance vs. AirTime
plt.subplot(1, 2, 2)
sns.scatterplot(data=df, x='Distance', y='AirTime', alpha=0.5, hue='UniqueCarrier', palette='flare')
plt.title('Distance vs. AirTime')
plt.xlabel('Distance (miles)')
plt.ylabel('AirTime (minutes)')
plt.grid(True, linestyle='--', alpha=0.6)

plt.tight_layout()
plt.show()
```

#### 2. Categorical vs. Numerical: Box Plots

Now, let's look at how arrival delays vary across different days of the week and unique carriers.

```python
plt.figure(figsize=(16, 6))

# Box Plot: DayOfWeek vs. ArrDelay
plt.subplot(1, 2, 1)
sns.boxplot(data=df, x='DayOfWeek', y='ArrDelay', palette='coolwarm')
plt.title('Arrival Delay by Day of Week')
plt.xlabel('Day of Week')
plt.ylabel('Arrival Delay (minutes)')
plt.ylim(0, 200) # Limit y-axis for better visibility of the main distribution
plt.grid(axis='y', linestyle='--', alpha=0.6)

# Box Plot: UniqueCarrier vs. ArrDelay
plt.subplot(1, 2, 2)
sns.boxplot(data=df, x='UniqueCarrier', y='ArrDelay', palette='rocket')
plt.title('Arrival Delay by Unique Carrier')
plt.xlabel('Unique Carrier')
plt.ylabel('Arrival Delay (minutes)')
plt.ylim(0, 200) # Limit y-axis for better visibility of the main distribution
plt.xticks(rotation=45, ha='right')
plt.grid(axis='y', linestyle='--', alpha=0.6)

plt.tight_layout()
plt.show()
```

### Multivariate Analysis

#### 1. Correlation Matrix of Numerical Variables

Let's visualize the correlation between some of the important numerical features using a heatmap. This helps in understanding linear relationships between multiple pairs of variables simultaneously.

```python
correlation_cols = ['DepTime', 'ArrTime', 'ActualElapsedTime', 'CRSElapsedTime', 'AirTime', 'ArrDelay', 'DepDelay', 'Distance', 'TaxiIn', 'TaxiOut']

# Calculate the correlation matrix
corr_matrix = df[correlation_cols].corr()

plt.figure(figsize=(10, 8))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt=".2f", linewidths=.5)
plt.title('Correlation Matrix of Key Numerical Variables')
plt.show()
```

#### 2. Pair Plot for Selected Numerical Variables

A pair plot creates a grid of scatterplots for each pair of variables and histograms/KDEs for individual variables, allowing a quick overview of relationships and distributions.

```python
selected_for_pairplot = ['DepDelay', 'ArrDelay', 'Distance', 'AirTime']
sns.pairplot(df[selected_for_pairplot])
plt.suptitle('Pair Plot of Selected Numerical Variables', y=1.02)
plt.show()
```

#### 3. Faceted Plots for Categorical and Numerical Relationships

We can use `seaborn.catplot` to explore how a numerical variable (e.g., Arrival Delay) varies across multiple categorical variables, such as 'DayOfWeek' and 'UniqueCarrier'.

```python
# Using catplot to show Arrival Delay distribution across DayOfWeek, separated by UniqueCarrier
sns.catplot(data=df, x='DayOfWeek', y='ArrDelay', col='UniqueCarrier', col_wrap=3, kind='box', height=4, aspect=1.2, palette='muted', col_order=df['UniqueCarrier'].value_counts().index)
plt.suptitle('Arrival Delay by Day of Week across Unique Carriers', y=1.02, fontsize=16)
plt.tight_layout(rect=[0, 0.03, 1, 0.98])
plt.show()
```
